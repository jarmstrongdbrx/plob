```sql
WITH REC AS (
  SELECT 
    P.Policy.`Policy.Policy`.PolicyNumber AS policy_number,
    P.Policy.`Policy.Policy`.PolicyKey AS policy_key,
    P.Policy.`Policy.Policy`.SourceSystemDateCreated AS source_system_date_created,
    P.Policy.`Policy.Policy`.StatusKey AS status_key,
    P.Policy.`Policy.Policy`.StartDate AS start_date,
    L.LINE_OF_INSURANCE_KEY AS line_of_insurance_key,
    L.COVERAGE_KEY AS coverage_key,
    L.AS_LOB_KEY AS as_lob_key,
    L.SUBLINE_KEY AS subline_key,
    L.INSURED_OBJECT_KEY AS insured_object_key
  FROM prod_lakehouse.dev_john_armstrong_policy.bronze_policy P
  JOIN prod_lakehouse.dev_john_armstrong_policy.v_xref_policy_levels L 
    ON P.Policy.`Policy.Policy`.PolicyKey = L.POLICY_KEY 
    AND CAST(P.Policy.`Policy.Policy`.StartDate AS timestamp) = L.START_DATE
  WHERE UPPER(L.LINE_OF_INSURANCE_KEY) = 'PROPERTY'
    AND P.Policy.`Policy.Policy`.PolicyNumber IN (
      'CPP20000047600',
      'CP20000026800',
      'CP20000034700',
      'CPP20000042700',
      'CPP20000082000'
    )
)
SELECT * FROM REC LIMIT 5
```
